```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import random
import pandas as pd
import time

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
reviews = []

url = "https://www.trustpilot.com/review/www.spotify.com?page="

for page_num in range(1, 65): 
    driver.get(f"{url}{page_num}")
    time.sleep(6)

    # reject Cookies
    try:
        reject_button = WebDriverWait(driver, 5).until(
            EC.element_to_be_clickable((By.XPATH, '//*[@id="onetrust-reject-all-handler"]')) # find the element
        )
        reject_button.click() # click the reject button
    except:
        pass

    # scroll the page
    time.sleep(random.uniform(5, 10))
    
    for i in range(1, 4):
        driver.execute_script(f"window.scrollTo(0, {i * 300});")
        time.sleep(random.uniform(1, 2))

    # find review card
    review_cards = driver.find_elements(By.XPATH, '//article')
    for card in review_cards:
        try:
            # date
            date = card.find_element(By.XPATH, './/time').get_attribute("datetime")
            # title
            title = card.find_element(By.XPATH, './/h2').text
            # content
            try:
                # extract without line break
                content_raw = card.find_element(By.XPATH, ".//p").text
                content = content_raw.replace('\n', ' ').replace('\r', ' ')
            except:
                content = "No Content"
            # rating
            try:
                rating = card.find_element(By.XPATH, ".//img[contains(@alt, 'Rated')]").get_attribute("alt")
            except:
                rating = "No Rating"
            
            reviews.append({
                "date": date,
                "title": title,
                "content": content,
                "rating": rating
            })
        except:
            continue 

    print(f"Page {page_num} complete, total {len(reviews)}")
    if len(reviews) >= 1200: break # break the loop until 1200 reviews have collected

    # detect log in window: manually sign in when sign in window pop-up
    # pause to log in until pressing ENTER to continue
    if "Sign in" in driver.page_source:
        print(f"⚠️ Page {page_num} requires login!")
        input("Press Enter when logged in.")

df = pd.DataFrame(reviews)
df.to_csv("spotify_reviews.csv", index=False, encoding='utf-8-sig')
driver.quit()
```
