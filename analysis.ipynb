### Topic Modelling (LDA)

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from collections import Counter

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

# add stopwords in the list
custom_stops = [
    'spotify', 'app', 'music', 'good', 'company', 'worst', 'great', 
    'bad', 'best', 'dont', 'terrible', 'song', 'year', 'experience', 
    'cant', 'get', 'platform', 'love', 'poor', 'used', 'many', 'ever', 
    'using', 'could', 'give', 'still', 'ive', 'use', 'every', 'never', 
    'would', 'one', 'time', 'user', 'version', 'another', 'avoid', 'way', 
    'want', 'make', 'really', 'back', 'even', 'actually', 'also', 'know', 
    'think', 'take', 'like', 'song', 'listen', 'play', 'try', 'get',
    'say', 'year', 'people', 'day', 'month', 'hour', 'need', 'work', 
    'new', 'keep', 'ask', 'find', 'hour', 'need', 'without', 'tell',
    'add', 'chat', 'thing', 'start', 'end', 'doesnt', 'minute', 'much',
    'see', 'didnt', 'well', 'wont'
    ]
stop_words.update(custom_stops)

# lemmatization
def lemmatize_and_filter(text):
    words = str(text).split()
    clean_words = [lemmatizer.lemmatize(w, pos='v') for w in words if w not in stop_words and len(w) > 2] # convert verbs into base form
    clean_words = [lemmatizer.lemmatize(w, pos='n') for w in clean_words] # convert plural into singular
    return clean_words

df['tokens'] = df['full_review_clean'].apply(lemmatize_and_filter)
df.head()
```

```python
# update stopwrods, same as the previous one
def lemmatize_and_filter(text):
    words = [w.strip() for w in str(text).split()]
    
    # verb and noun
    lemmatized_words = []
    for w in words:
        v_lemma = lemmatizer.lemmatize(w, pos='v')
        n_lemma = lemmatizer.lemmatize(v_lemma, pos='n')
        lemmatized_words.append(n_lemma)
    
    # convert speech
    clean_words = [w for w in lemmatized_words if w not in stop_words and len(w) > 2]
    
    return clean_words

df['tokens'] = df['full_review_clean'].apply(lemmatize_and_filter)

all_words = [word for tokens in df['tokens'] for word in tokens]
top_30 = Counter(all_words).most_common(30)
```

```python
import gensim
from gensim import corpora

dictionary = corpora.Dictionary(df['tokens'])

# delete words contained below 5 comments or more than 50% comments
dictionary.filter_extremes(no_below=5, no_above=0.5)

# count the frequency of the words in every comments
corpus = [dictionary.doc2bow(text) for text in df['tokens']]
```

```python
# find the optimal k value to decide the number of topic
from gensim.models.coherencemodel import CoherenceModel

def compute_coherence_values(dictionary, corpus, texts, limit, start=2, step=1):
    coherence_values = []
    model_list = []
    # run topic number in a loop
    for num_topics in range(start, limit, step):
        model = gensim.models.LdaModel(corpus=corpus, id2word=dictionary, num_topics=num_topics, random_state=42)
        model_list.append(model)
        # count the score
        coherencemodel = CoherenceModel(model=model, texts=texts, dictionary=dictionary, coherence='c_v')
        coherence_values.append(coherencemodel.get_coherence())

    return model_list, coherence_values

# test k
model_list, coherence_values = compute_coherence_values(dictionary=dictionary, corpus=corpus, texts=df['tokens'], start=2, limit=11, step=1)

x = range(2, 11, 1)
plt.plot(x, coherence_values)
plt.xlabel("Number of Topics")
plt.ylabel("Coherence Score")
plt.legend(("coherence_values"), loc='best')
plt.show()
```

```python
# set the number of topic
num_topics = 4

lda_model = gensim.models.LdaModel(
    corpus=corpus,
    id2word=dictionary,
    num_topics=num_topics, 
    random_state=42,
    passes=15,
    alpha='auto'
)

# print the keywords of every topic
for idx, topic in lda_model.print_topics(-1):
    print(f"Topic: {idx} \nWords: {topic}\n")
```

### Sentiment Analysis

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

# sentiment score
df['sentiment_score'] = df['full_review_clean'].apply(lambda x: analyzer.polarity_scores(str(x))['compound'])
# overall average sentiment score
print(f"Overall Average Sentiment Score: {df['sentiment_score'].mean():.2f}")
```

```python
# get the topic
def get_dominant_topic(tokens):
    bow = dictionary.doc2bow(tokens) # format
    topics = lda_model.get_document_topics(bow) # get the distribution of the possibility in every topic 
    dominant = sorted(topics, key=lambda x: x[1], reverse=True)[0][0] # find the highest possibility
    return dominant

# assign topic
df['topic'] = df['tokens'].apply(get_dominant_topic)
```

```python
# name the topics
topic_mapping = {
    0: "Artist & Content",
    1: "Comparative Service & Ad-Experience",
    2: "Billing & Subscription Management",
    3: "Technical Issues & Account Access Support"
}

df['topic_name'] = df['topic'].map(topic_mapping)

# average sentiment score of each topic
final_analysis = df.groupby('topic_name')['sentiment_score'].mean().sort_values()

print("------ Average Sentiment Score of Topic ------")
print(final_analysis)
```

### Word Cloud

```python
from wordcloud import WordCloud

# plot the word cloud
def create_topic_wordcloud(topic_label, color_map):
    topic_words = [word for tokens in df[df['topic_name'] == topic_label]['tokens'] for word in tokens]
    text = " ".join(topic_words)
    
    wordcloud = WordCloud(
        width=800, 
        height=400, 
        background_color='white',
        colormap=color_map,
        max_words=100,
        collocations=False
    ).generate(text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.title(f"{topic_label}", fontsize=15, pad=20)
    plt.axis('off')
    plt.show()

topic_colors = {
    "Billing & Subscription Management": "Reds",
    "Technical Issues & Account Access Support": "Blues",
    "Artist & Content": "Greens",
    "Comparative Service & Ad-Experience": "Oranges"
}

# plot every topic with a loop
for name, color in topic_colors.items():
    if name in df['topic_name'].unique():
        create_topic_wordcloud(name, color)
```
